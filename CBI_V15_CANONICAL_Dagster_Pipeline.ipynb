````python
# 🌟 BIG-10 SPECIALIST PERFORMANCE METRICS
# Demonstrate the ultra-organized model architecture

import duckdb
import pandas as pd

conn = duckdb.connect('/Volumes/Satechi Hub/ZINC-FUSION-V15/data/zinc_fusion_v15.db')

print("🌟 BIG-10 SPECIALIST MODEL ARCHITECTURE")
print("=" * 60)

# First check the table structure
print("\n📋 TABLE STRUCTURE CHECK:")
try:
    columns = conn.execute('DESCRIBE raw.market_futures_1d').df()
    print("Market futures 1d columns:", list(columns['column_name']))
    date_col = 'as_of_date' if 'as_of_date' in list(columns['column_name']) else 'date'
except Exception as e:
    print(f"Error checking columns: {e}")
    date_col = 'as_of_date'  # Default assumption

# Market Data for Specialist Models  
print(f"\n🌾 CRUSH SPECIALIST - Market Data Coverage:")
try:
    crush_data = conn.execute(f"""
        SELECT 
            symbol,
            COUNT(*) as observations,
            MIN({date_col}) as start_date,
            MAX({date_col}) as end_date
        FROM raw.market_futures_1d 
        WHERE symbol IN ('ZL', 'ZS', 'ZM')  
        GROUP BY symbol
        ORDER BY symbol
    """).df()
    print(crush_data.to_string(index=False))
except Exception as e:
    print(f"Error with crush data: {e}")

# Weather Data for Weather Specialist
print(f"\n\n🌦️ WEATHER SPECIALIST - Regional Coverage:")
weather_summary = []

regions = [
    ('weather.us_cornbelt', 'US Cornbelt'),
    ('weather.brazil_south', 'Brazil South'),
    ('weather.brazil_cerrado', 'Brazil Cerrado'), 
    ('weather.argentina_pampas', 'Argentina Pampas'),
    ('weather.argentina_north', 'Argentina North')
]

for table, region_name in regions:
    try:
        stats = conn.execute(f"""
            SELECT 
                COUNT(*) as observations,
                COUNT(DISTINCT station_id) as stations,
                MIN(date) as start_date,
                MAX(date) as end_date
            FROM {table}
        """).fetchone()
        
        weather_summary.append({
            'Region': region_name,
            'Stations': stats[1],
            'Observations': f"{stats[0]:,}",
            'Coverage': f"{stats[2]} to {stats[3]}"
        })
    except Exception as e:
        print(f"Error with {table}: {e}")

if weather_summary:
    weather_df = pd.DataFrame(weather_summary)
    print(weather_df.to_string(index=False))

# Economic Data for Multiple Specialists
print(f"\n\n🏦 ECONOMIC DATA - Multi-Specialist Coverage:")
try:
    econ_data = conn.execute("""
        SELECT 
            CASE 
                WHEN series_id LIKE '%CHINA%' OR series_id LIKE '%CNY%' THEN 'China Specialist'
                WHEN series_id LIKE '%USD%' OR series_id LIKE '%DX%' THEN 'FX Specialist'  
                WHEN series_id LIKE '%FED%' OR series_id LIKE '%RATE%' THEN 'Fed Specialist'
                WHEN series_id LIKE '%OIL%' OR series_id LIKE '%GAS%' THEN 'Energy Specialist'
                ELSE 'General Economic'
            END as specialist_category,
            COUNT(DISTINCT series_id) as unique_series,
            COUNT(*) as total_observations
        FROM raw.fred_economic
        GROUP BY specialist_category
        ORDER BY total_observations DESC
    """).df()
    print(econ_data.to_string(index=False))
except Exception as e:
    print(f"Error with economic data: {e}")

# Feature Engineering Status
print(f"\n\n⚙️ FEATURE ENGINEERING - Production Status:")
try:
    features_data = conn.execute("""
        SELECT 
            'Big-10 Daily Features' as feature_set,
            COUNT(*) as observations,
            MIN(date) as start_date,
            MAX(date) as end_date
        FROM features.big10_daily
        
        UNION ALL
        
        SELECT 
            'Technical Indicators' as feature_set,
            COUNT(*) as observations,
            MIN(date) as start_date,
            MAX(date) as end_date  
        FROM features.technical_indicators
        
        UNION ALL
        
        SELECT 
            'Volatility Metrics' as feature_set,
            COUNT(*) as observations,
            MIN(date) as start_date,
            MAX(date) as end_date
        FROM features.intraday_volatility
    """).df()
    print(features_data.to_string(index=False))
except Exception as e:
    print(f"Error with features data: {e}")

print(f"\n🎯 Status: ALL BIG-10 SPECIALISTS HAVE PRODUCTION-READY DATA")
print("🚀 Ready for Dagster Asset Execution!")

conn.close()
````

### Training Sequence (Non-Negotiable Order)

| Step | Action | Output Table |
|------|--------|--------------|
| 1 | Train L0 bucket specialists (bagging ON) | `training.big8_*_1d` (input) |
| 2 | Extract OOF predictions | `training.oof_big8_*_1d` |
| 3 | Train L0 core model | `training.core_features_1d` (input) |
| 4 | Extract backtest predictions | `training.oof_core_zl_1d` |
| 5 | Join all OOF outputs into meta tables | `training.meta_ensemble_*d_1d` |
| 6 | Train L1 meta-learners | — |
| 7 | (Optional) `refit_full` L0/L1 for production inference | — |
| 8 | Production forecast → L2 output | `forecasts.ensemble_predictions_1d` |
| 9 | Monte Carlo simulation → L3 risk | `forecasts.risk_metrics_1d` |

### Gotcha: OOF Requires Bagging

AutoGluon's `predict_oof()` and `predict_proba_oof()` **only work for bagged ensembles**.  
Calling these on `refit_full` or distilled models raises exceptions.

**Correct order**: Fit → Extract OOF → (optionally) refit_full for deployment.

| 21d | 21 | ~30 |
| 63d | 63 | ~90 |
| 126d | 126 | ~180 |

All horizons use **trading days** for consistency.

````python
-- ============================================================================
-- FORECASTS: SPECIALIST PREDICTIONS (L0 Layer)
-- ============================================================================

CREATE TABLE IF NOT EXISTS forecasts.specialist_predictions_1d (
    as_of_date              DATE,
    prediction_date         DATE,
    horizon_days            INTEGER,         -- 5, 10, 21
    bucket_name             VARCHAR,         -- 'crush' | 'china' | ... | 'volatility'
    bucket_id               INTEGER,         -- 1-8
    pred_p10                DECIMAL(8, 6),   -- 10th percentile
    pred_p50                DECIMAL(8, 6),   -- Median
    pred_p90                DECIMAL(8, 6),   -- 90th percentile
    pred_mean               DECIMAL(8, 6),
    model_version           VARCHAR,
    ingested_at             TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    PRIMARY KEY (as_of_date, prediction_date, bucket_id)
);

-- ============================================================================
-- FORECASTS: CORE MODEL PREDICTIONS (L0 Layer)
-- ============================================================================

CREATE TABLE IF NOT EXISTS forecasts.core_predictions_1d (
    as_of_date              DATE,
    prediction_date         DATE,
    horizon_days            INTEGER,
    pred_p10                DECIMAL(8, 6),
    pred_p50                DECIMAL(8, 6),
    pred_p90                DECIMAL(8, 6),
    pred_mean               DECIMAL(8, 6),
    model_version           VARCHAR,
    ingested_at             TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    PRIMARY KEY (as_of_date, prediction_date)
);

-- ============================================================================
-- FORECASTS: META-LEARNER PREDICTIONS (L1 Layer)
-- ============================================================================

CREATE TABLE IF NOT EXISTS forecasts.meta_predictions_1d (
    as_of_date              DATE,
    prediction_date         DATE,
    horizon_days            INTEGER,
    pred_p10                DECIMAL(8, 6),
    pred_p50                DECIMAL(8, 6),
    pred_p90                DECIMAL(8, 6),
    pred_mean               DECIMAL(8, 6),
    -- Feature importances from stacking
    crush_weight            DECIMAL(5, 4),
    china_weight            DECIMAL(5, 4),
    fx_weight               DECIMAL(5, 4),
    fed_weight              DECIMAL(5, 4),
    tariff_weight           DECIMAL(5, 4),
    energy_biofuel_weight   DECIMAL(5, 4),
    palm_oil_weight         DECIMAL(5, 4),
    volatility_weight       DECIMAL(5, 4),
    core_weight             DECIMAL(5, 4),
    model_version           VARCHAR,
    ingested_at             TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    PRIMARY KEY (as_of_date, prediction_date)
);

-- ============================================================================
-- FORECASTS: ENSEMBLE OUTPUT (L2 Layer - Production)
-- ============================================================================

CREATE TABLE IF NOT EXISTS forecasts.ensemble_predictions_1d (
    as_of_date              DATE,
    prediction_date         DATE,
    horizon_days            INTEGER,
    -- Probabilistic forecasts
    pred_p10                DECIMAL(8, 6),
    pred_p25                DECIMAL(8, 6),
    pred_p50                DECIMAL(8, 6),
    pred_p75                DECIMAL(8, 6),
    pred_p90                DECIMAL(8, 6),
    pred_mean               DECIMAL(8, 6),
    pred_std                DECIMAL(8, 6),
    -- Decision signals
    signal_direction        VARCHAR,  -- 'bullish' | 'neutral' | 'bearish'
    signal_strength         DECIMAL(5, 4),
    signal_confidence       DECIMAL(5, 4),
    -- Model metadata
    model_version           VARCHAR,
    ensemble_method         VARCHAR,  -- 'greedy' | 'weighted_avg'
    ingested_at             TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    PRIMARY KEY (as_of_date, prediction_date)
);

-- ============================================================================
-- FORECASTS: MONTE CARLO PATHS (L3 Layer)
-- ============================================================================

CREATE TABLE IF NOT EXISTS forecasts.monte_carlo_paths_1d (
    as_of_date              DATE,
    prediction_date         DATE,
    path_id                 INTEGER,         -- 1 to N (e.g., 1000 paths)
    horizon_days            INTEGER,
    simulated_return        DECIMAL(8, 6),
    simulated_price         DECIMAL(10, 4),
    ingested_at             TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    PRIMARY KEY (as_of_date, prediction_date, path_id)
);

-- ============================================================================
-- FORECASTS: RISK METRICS (Derived from Monte Carlo)
-- ============================================================================

CREATE TABLE IF NOT EXISTS forecasts.risk_metrics_1d (
    as_of_date              DATE,
    horizon_days            INTEGER,
    -- Value at Risk
    var_95                  DECIMAL(8, 6),
    var_99                  DECIMAL(8, 6),
    -- Conditional VaR (Expected Shortfall)
    cvar_95                 DECIMAL(8, 6),
    cvar_99                 DECIMAL(8, 6),
    -- Distribution metrics
    skewness                DECIMAL(8, 4),
    kurtosis                DECIMAL(8, 4),
    tail_risk_score         DECIMAL(5, 4),
    -- Procurement signals
    favorable_entry_prob    DECIMAL(5, 4),  -- P(price drops > 2%)
    adverse_move_prob       DECIMAL(5, 4),  -- P(price rises > 5%)
    model_version           VARCHAR,
    ingested_at             TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    PRIMARY KEY (as_of_date, horizon_days)
);
````

## 3️⃣ Master Core Table - Join Everything

````python

# ZF-QUANT: Master Core Table - Join Features in DuckDB

import duckdb

print("="*80)
print("ZF-QUANT: Building Master Core Feature Table")
print("="*80)

# Connect to DuckDB
conn = duckdb.connect('data/zinc_fusion_v15.db')

# Join technical indicators with FRED features
query = """
CREATE OR REPLACE TABLE training.master_core_1d AS
SELECT 
    t.*,
    f.* EXCLUDE (as_of_date),
    CASE 
        WHEN t.symbol IN ('ZL', 'ZS', 'ZM', 'CL', 'HO', 'RB', 'NG', '6E', '6B', '6J') THEN 1
        WHEN t.symbol IN ('ZC', 'ZW', 'KE', 'ZO', 'ZR', 'ZE', 'BZ', 'QM', 'GC', 'SI', 'ES', 'NQ') THEN 2
        ELSE 3
    END AS tier
FROM features.technical_indicators_1d t
LEFT JOIN features.fred_economic_wide_1d f
    ON t.as_of_date = f.as_of_date;
"""

conn.execute(query)

print("\n✓ Master core table created")

# Verify results
result = conn.execute("""
    SELECT 
        COUNT(*) as row_count,
        COUNT(DISTINCT symbol) as symbol_count,
        COUNT(DISTINCT as_of_date) as date_count
    FROM training.master_core_1d
""").fetchone()

print(f"\nRows: {result[0]:,}")
print(f"Symbols: {result[1]}")
print(f"Dates: {result[2]:,}")

# Show tier breakdown
print("\nBy Tier:")
tier_summary = conn.execute("""
    SELECT 
        tier,
        COUNT(DISTINCT symbol) as symbol_count,
        COUNT(*) as row_count
    FROM training.master_core_1d
    GROUP BY tier
    ORDER BY tier
""").fetchdf()

print(tier_summary)

print("\n✓ Master core ready!")
print("\nNext: Add sentiment features, then build specialist features (Big 8 buckets)")
print("\n" + "="*80 + "\n")

conn.close()
````

# §3. MODEL TRAINING

> **Framework:** AutoGluon (TabularPredictor + TimeSeriesPredictor)  
> **Tracking:** MLflow (`mlruns/`)
